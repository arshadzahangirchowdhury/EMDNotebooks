# Design Safety Factor Selection

```python
import ipywidgets as widgets
from IPython.display import display
```

```python
factors = [
    "1. Accuracy of loads",
    "2. Accuracy of stress calculation",
    "3. Accuracy of strength",
    "4. Need to conserve mat'ls/weight/cost",
    "5. Seriousness of failure",
    "6. Quality of workmanship",
    "7. Conditions of operation",
    "8. Quality of inspection/maintenance",
]

rn_inputs = [widgets.BoundedFloatText(value=0.0, min=-4.0, max=4.0, step=1, description=f[:2], layout=widgets.Layout(width="220px")) for f in factors]

rows = [widgets.HBox([widgets.Label(f, layout=widgets.Layout(width="320px")), rn]) for f, rn in zip(factors, rn_inputs)]
display(widgets.VBox(rows))



def n_d(t):
    if t >= -6:
        return 1 + (10 + t) ** 2 / 100
    return 1.15

out = widgets.Output()

def compute(_=None):
    t = sum(rn.value for rn in rn_inputs)
    with out:
        out.clear_output()
        print(f"t = {t:.2f}")
        print(f"n_d = {n_d(t):.4f}")

btn = widgets.Button(description="Compute")
btn.on_click(compute)
display(btn, out)
```

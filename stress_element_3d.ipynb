```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
import ipywidgets as widgets
from IPython.display import display
```

```python
# unit cube centered at origin, edges length 2
r = 1
verts = np.array([[x, y, z] for x in (-r, r) for y in (-r, r) for z in (-r, r)])
edges = [(0,1),(0,2),(0,4),(1,3),(1,5),(2,3),(2,6),(3,7),(4,5),(4,6),(5,7),(6,7)]

# stress vectors: origin + direction, one pair per face (normal in, normal out) plus shear pairs
face_normals = {
    "+x": np.array([1,0,0]), "-x": np.array([-1,0,0]),
    "+y": np.array([0,1,0]), "-y": np.array([0,-1,0]),
    "+z": np.array([0,0,1]), "-z": np.array([0,0,-1]),
}

def Rx(t):
    c, s = np.cos(t), np.sin(t)
    return np.array([[1,0,0],[0,c,-s],[0,s,c]])

def Ry(t):
    c, s = np.cos(t), np.sin(t)
    return np.array([[c,0,s],[0,1,0],[-s,0,c]])

def Rz(t):
    c, s = np.cos(t), np.sin(t)
    return np.array([[c,-s,0],[s,c,0],[0,0,1]])
```

```python
def plot_element(theta_x, theta_y, theta_z):
    R = Rz(np.radians(theta_z)) @ Ry(np.radians(theta_y)) @ Rx(np.radians(theta_x))
    v = verts @ R.T

    fig = plt.figure(figsize=(6,6))
    ax = fig.add_subplot(111, projection="3d")

    for i, j in edges:
        ax.plot(*zip(v[i], v[j]), color="k")

    for face, n in face_normals.items():
        origin = (n * r) @ R.T
        d = n @ R.T
        ax.quiver(*origin, *d, length=1.0, color="r", arrow_length_ratio=0.2)

    lim = 2.2
    ax.set_xlim(-lim, lim); ax.set_ylim(-lim, lim); ax.set_zlim(-lim, lim)
    ax.set_xlabel("x"); ax.set_ylabel("y"); ax.set_zlabel("z")
    ax.set_box_aspect([1,1,1])
    plt.show()

widgets.interact(
    plot_element,
    theta_x=widgets.IntSlider(value=0, min=-180, max=180, step=5, description="theta_x"),
    theta_y=widgets.IntSlider(value=0, min=-180, max=180, step=5, description="theta_y"),
    theta_z=widgets.IntSlider(value=0, min=-180, max=180, step=5, description="theta_z"),
)
```

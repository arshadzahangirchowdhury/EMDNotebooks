```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display
```

```python
x = np.linspace(-20, 20, 1000)

def plot_normal(mu, sigma):
    y = 1 / (sigma * np.sqrt(2 * np.pi)) * np.exp(-0.5 * ((x - mu) / sigma) ** 2)
    plt.figure(figsize=(7, 4))
    plt.plot(x, y)
    plt.xlim(-20, 20)
    plt.ylim(0, 1)
    plt.title(f"mu={mu}, sigma={sigma}")
    plt.xlabel("x")
    plt.ylabel("f(x)")
    plt.grid(True)
    plt.show()

widgets.interact(
    plot_normal,
    mu=widgets.FloatSlider(value=0, min=-10, max=10, step=0.5),
    sigma=widgets.FloatSlider(value=1, min=0.1, max=5, step=0.1),
)
```
